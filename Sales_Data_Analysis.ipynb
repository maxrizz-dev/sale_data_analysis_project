#SECTION C - CLEANING THE DATASET

```python
df.isnull().sum()
```


 Drop fully-empty columns

```python
df.drop(['Status','unnamed1'], axis=1, inplace=True)
```

 Drop rows where Amount is missing


```python
df.dropna(subset=['Amount'], inplace=True)
```

checking for duplicates value

```python
df_dup = df.duplicated().sum()
```

```python
df_dup
```

```python
df_cleaned = df.drop_duplicates()
```

```python
new_dup_count = df_cleaned.duplicated().sum()
print(new_dup_count)
```

incorrect data type

```python
df['Amount'] = df['Amount'].astype(int)
```

#SECTION D - EXPLORATORY DATA ANALYSIS

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Age'], df['Amount'], alpha=0.5, c=df['Amount'], cmap='autumn')
plt.title("Age vs Investment")
plt.xlabel("Age")
plt.ylabel("Investment Amount")
plt.colorbar(label='Investment Amount')

plt.show()
```

```python
age_investment = df.groupby('Age')['Amount'].sum()

plt.figure(figsize=(9,5))
plt.hist(df['Age'], bins=10, color='skyblue', edgecolor='black')
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
gender_amount = df.groupby("Gender")["Amount"].sum()

plt.pie(gender_amount.values,
        labels=gender_amount.index,
        autopct="%1.1f%%")

plt.title("Amount by Gender")

plt.show()
```

```python
occupation_investment = df.groupby('Occupation')['Amount'].sum()
top_occupation = occupation_investment.sort_values(ascending=False).head(10)

plt.figure(figsize=(8,8))
top_occupation.plot(
    kind='pie',
    autopct='%1.1f%%',
    radius=0.8,
    startangle=140,
    colors=plt.cm.Paired.colors,
    rotatelabels=True
)
plt.title("Top 10 Occupation Investment Share")
plt.ylabel("")
plt.show()
```

```python
state_orders = df.groupby('State')['Amount'].sum()
plt.bar(state_orders.index, state_orders.values, color='coral', label='Total Investment')
plt.xticks(rotation=90)
plt.title("Total Investment by State")
plt.xlabel("State")
plt.ylabel("Total Investment")
plt.legend()
plt.show()
```

```python
age_count = df['Age'].value_counts().sort_index()

plt.figure(figsize=(12,6))
age_count.plot(kind='bar', color='lightgreen')
plt.title("Number of Orders by Age")
plt.xlabel("Number of Orders")
plt.ylabel("Age")
plt.show()
```
